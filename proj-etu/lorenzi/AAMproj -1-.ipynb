## Exercice 1
On se place sur $X = {\mathbf{R}}^3_+$ et on considère la relation de préférence notée $\preccurlyeq$ définie comme suit : 
$(x_1,x_2,x_3) \preccurlyeq (y_1,y_2,y_3)$ si et seulement pour au moins deux valeurs de $i$ dans $\{1, 2, 3\}$ on a $x_i \leq y_i$.

### 1.1
Montrer que cette relation de préférence est complète.    

**Réponse :**      
Une relation de préférence est dite complète si tous les éléments de $X = {\mathbf{R}}^3_+$ sont comparables, c'est-à-dire: $ \forall (x,y) \in X^2, x \preccurlyeq y $ ou $ y \preccurlyeq x$.$$
$$ Soient $(x_1, x_2, x_3)$ et $(y_1, y_2, y_3) \in X^2$, détaillons les quatre cas qui montrent que $(x_1, x_2, x_3) \preccurlyeq (y_1, y_2, y_3)$ et les quatre cas complémentaires pour $(y_1, y_2, y_3) \preccurlyeq (x_1, x_2, x_3)$. 

*   $x_1 \preccurlyeq y_1, x_2 \preccurlyeq y_2, x_3 \preccurlyeq y_3$
*   $x_1 \succcurlyeq y_1, x_2 \preccurlyeq y_2, x_3 \preccurlyeq y_3$
*   $x_1 \preccurlyeq y_1, x_2 \succcurlyeq y_2, x_3 \preccurlyeq y_3$
*   $x_1 \preccurlyeq y_1, x_2 \preccurlyeq y_2, x_3 \succcurlyeq y_3$
$$
$$
*   $y_1 \preccurlyeq x_1, y_2 \preccurlyeq x_2, y_3 \preccurlyeq x_3$
*   $y_1 \succcurlyeq x_1, y_2 \preccurlyeq x_2, y_3 \preccurlyeq x_3$
*   $y_1 \preccurlyeq x_1, y_2 \succcurlyeq x_2, y_3 \preccurlyeq x_3$
*   $y_1 \preccurlyeq x_1, y_2 \preccurlyeq x_2, y_3 \succcurlyeq x_3$

### 1.2
À l'aide d'un contrexemple, montrer que cette relation de préférence n'est pas transitive.

**Réponse :**

Une relation de préférence est dite transitive si $\forall (x,y,z) ∈ X^3$, $x \preccurlyeq y$ et $y \preccurlyeq z \Rightarrow x \preccurlyeq z$ . 

Prenons $x = (x_1,x_2,x_3); y = (y_1,y_2,y_3); z = (z_1,z_2,z_3)$ $$
$$
$$
\left\{
\begin{array}{c}
x_1 \preccurlyeq y_1\\
x_2 \preccurlyeq y_2\\
x_3 \succcurlyeq y_3\\
\end{array}
\right. \Rightarrow x \preccurlyeq y \qquad \qquad \left\{
\begin{array}{c}
y_1 \succcurlyeq z_1\\
y_2 \preccurlyeq z_2\\
y_3 \preccurlyeq z_3\\
\end{array}
\right. \Rightarrow y \preccurlyeq z
$$ $$
$$
Conclusion : $x_2 \preccurlyeq y_2 \preccurlyeq z_2$ $\quad$ $z_1 \preccurlyeq x_1 \preccurlyeq y_1$ $\quad$ $y_3 \preccurlyeq z_3 \preccurlyeq x_3$ $\quad$ donc z $\preccurlyeq$ x. $$
$$Au final, cette relation de préférence n'est pas transitive.

## Exercice 3B
On considère le jeu de poker simplifié à deux joueurs suivant. Trois mains sont possibles, numérotées de $1$ à $3$, de la plus faible à la plus forte. Chaque joueuse, ayant reçu l'une de ces mains, met un jeton au pot. Le jeu se déroule ensuite selon le schéma ci-dessous (jeu séquentiel), sachant que, pour _miser_ ou pour _voir_, une joueuse doit rajouter un jeton au pot (rien à mettre au pot si la joueuse choisit de _passer_). La gagnante remporte le pot. 
La joueuse 2 triche et, à l'insu de la joueuse 1, s'est arrangée avec le croupier qui distribue les cartes afin que la main $(1, 3)$ (qui lui est favorable) sorte plus souvent que les autres, avec une probabilité de $1/4$ (les cinq autres mains sortant avec une même probabilité de $3/20$).

### 3.1
Calculer la matrice des gains (en espérance) du jeu sous forme normale.

**Réponse:**

La matrice des gains correspond à la matrice A. Elle est représentée par le code ci-dessus: 



````python
 import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """


# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 
 
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
        if i == 1:
          g = g + (1/4) * gmfun(s1, s2, Mains[i])
        else:
          g = g + (3/20) * gmfun(s1, s2, Mains[i])
    return g

A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
        A[i, j] = gfun(S1[i], S2[j])

d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte


print('A = ', A)
````

### 3.4
Déterminer la valeur du jeu en stratégies mixtes, et donner un couple de stratégies mixtes en équilibre.

**Réponse:**

Une stratégie mixte pour un joueur est une distribution de probabilité sur l'ensemble de ses stratégies pures. 

Dans un équilibre en stratégies mixtes, chacun des joueurs est indifférent entre jouer sa stratégie mixte ou jouer l'une de ses stratégies pures, qui a été utilisé pour sa stratégie mixte. 


````python
import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """

class PbResolution(Exception):
    """Probleme de resolution du LP
    """

# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 
 
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
        if i == 1:
          g = g + (1/4) * gmfun(s1, s2, Mains[i])
        else:
          g = g + (3/20) * gmfun(s1, s2, Mains[i])
    return g

#Matrice des gains
A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
        A[i, j] = gfun(S1[i], S2[j])

d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte


# Calcul de la valeur du jeu en stratégies mixtes (minimax)
c =-np.ones(n) # min (c|x) = -max (-c|x)
b = np.ones(m) # Ax <= b 
bounds = (0, None) # 0 <= x

method='simplex'
#method='revised simplex'
#method='interior-point'

res = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method=method)
if not res.success: raise PbResolution

v = 1 / (-res.fun)
x = v * res.x
v = v - d
xx = np.round_(x, decimals=8)

print('v = ', v)
print('x = ', xx)
print('')
print("Stratégie mixte optimale de la joueuse 2 :")
for j in range(0, n):
    if xx[j] > 0:
        print(S2[j], "avec proba", xx[j])

        
````


### 3.5
Un nouveau croupier remplace le précédent. Quatre mains sont désormais possibles, numérotées de $1$ à $4$, de la plus faible à la plus forte. Donner l'ensemble des stratégies (pures) pour chaque joueuse dans le jeu sous forme normale.

**Réponse:**
La forme normale est utilisée afin de décrire des jeux à nombres de coups, de joueurs et de stratégies finis.$$
$$ Soit donc un ensemble fini de joueuses *J = {1,2}*. Chaque joueuse 1 ou 2 peut employer un nombre fini de stratégies pures choisies dans l'ensemble des possibilités ou choix qui lui sont disponibles.$$
$$ $\rightarrow$ Ici, $S_1$ est l'ensemble des stratégies de la joueuse 1, sachant qu'une stratégie détermine à l'avance ce que la joueuse 1 joue dans chacun des cas possibles. Connaissant sa main, la joueuse 1 peut choisir:

* soit de miser (M)
* soit de passer puis de voir (PV)
* soit de passer puis de passer (PP)$$
$$ $\Rightarrow$ 3 choix possibles: (M, PV, PP)$$
$$Ce choix dépend bien sûr de la main de la joueuse 1. Une stratégie consiste donc à déterminer pour chacune des mains possibles ce choix:
$$ S_1 = \{M, PV,PP\} \times \{M, PV,PP\} \times \{M, PV,PP\} \times \{M, PV,PP\}$$


$\rightarrow$ Ici, $S_2$ est une stratégie qui consiste à prévoir ce que la joueuse 2 joue, connaissant sa main, dans chacun des états possibles (c'est-à-dire selon le choix de la joueuse 1):
