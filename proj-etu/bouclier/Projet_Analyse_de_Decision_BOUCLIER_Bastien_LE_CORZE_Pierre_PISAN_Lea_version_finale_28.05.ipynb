## Exercice 1
*On se place sur* $X = {\mathbf{R}}^3_+$ *et on considère la relation de préférence notée* $\preccurlyeq$ *définie comme suit :* 
$(x_1,x_2,x_3) \preccurlyeq (y_1,y_2,y_3)$ *si et seulement pour au moins deux valeurs de* $i$ *dans* $\{1, 2, 3\}$ *on a* $x_i \leq y_i$.

###1.1 
*Montrer cette relation de préférence est complète.*

* Une relation de préférence est dite complète si tout les élément dans $X = {\mathbf{R}}^3_+$ sont comparable : 

 $\forall \;\; (x_1,x_2,x_3),(y_1,y_2,y_3) \in X^2$ : 
 
 $(x_1,x_2,x_3)\preccurlyeq (y_1,y_2,y_3)$ ou 
 $(y_1,y_2,y_3)\preccurlyeq (x_1,x_2,x_3)$
 

* Nous allons montrer que $(x_1,x_2,x_3)\preccurlyeq (y_1,y_2,y_3)$ ou 
  $(y_1,y_2,y_3)\preccurlyeq (x_1,x_2,x_3)$

  - si $x_1\preccurlyeq y_1,\; x_2 \preccurlyeq y_2 ,\; x_3\preccurlyeq y_3~~$: 
> $(x_1,x_2,x_3)\preccurlyeq(y_1,y_2,y_3)$

  - si $x_1 \succcurlyeq y_1 ,\; x_2 \succcurlyeq  y_2 ,\;  x_3 \succcurlyeq y_3$ :
> $(y_1,y_2,y_3)\preccurlyeq(x_1,x_2,x_3)$

  - si $x_1\preccurlyeq y_1 ,\; x_2 \preccurlyeq y_2 ,\; x_3 \succcurlyeq y_3$ 
  ou $x_2\preccurlyeq y_2 ,\; x_3 \preccurlyeq y_3 ,\; x_1 \succcurlyeq y_1$  ou $x_3\preccurlyeq y_3 ,\; x_1 \preccurlyeq y_1 ,\; x_2 \succcurlyeq y_2$ :
> $(x_1,x_2,x_3)\preccurlyeq(y_1,y_2,y_3)$

  - si $y_1\preccurlyeq x_1 ,\; y_2 \preccurlyeq x_2 ,\; y_3\succcurlyeq x_3$ ou $y_2\preccurlyeq x_2 ,\; y_3 \preccurlyeq x_3 ,\; y_1 \succcurlyeq x_1$  ou $y_3\preccurlyeq x_3 ,\; y_1 \preccurlyeq x_1 ,\; y_2 \succcurlyeq x_2$ :
>  $(y_1,y_2,y_3)\preccurlyeq(x_1,x_2,x_3)$

  $\Rightarrow$ Donc cette relation de préférence est complète.

### 1.2
*À l'aide d'un contrexemple, montrer que cette relation de préférence n'est pas transitive.*

* Soit $\preccurlyeq$ une relation de préférence, on dit que cette relation est transitive si : 

 $\forall \;\; (x_1,x_2,x_3),(y_1,y_2,y_3),(z_1,z_2,z_3) \in X^3$ :

 $(x_1x_2,x_3)\preccurlyeq(y_1,y_2,y_3)\;$ et $\;(y_1,y_2,y_3)\preccurlyeq(z_1,z_2,z_3)$
 
 $\Rightarrow$ $(x_1,x_2,x_3) \preccurlyeq (z_1,z_2,z_3)$

* Nous allons montrer que la relation de préférence n'est pas transitive, c'est à dire:

  $\exists \; [(x_1,x_2,x_3),\;(y_1,y_2,y_3),\;(z_1,z_2,z_3)] \in X^3$

  $(x_1,x_2,x_3) \preccurlyeq \;(y_1,y_2,y_3)\;$ et 
  
  $\;(y_1,y_2,y_3)\preccurlyeq \; (z_1,z_2,z_3)\;$ et
  
  $(z_1,z_2,z_3) \preccurlyeq \; (x_1,x_2,x_3)$

 >Par exemple, prenons:
 
   $\;\;x_1=1$ , $\;\;x_2=2$, $\;\;x_3=6$ et $\;\;y_1=4$, $\;\;y_2=3$, $\;\;y_3=2$ :
   
   on a $\;\;x_1 \preccurlyeq y_1$, $\;\;x_2 \preccurlyeq y_2$ et $\;\;y_3 \preccurlyeq x_3$ donc $(x_1,x_2,x_3)\preccurlyeq (y_1,y_2,y_3)$

 Ici si l'on prend: $\;\; z_1=5$, $\;\; z_2=1$, $\;\; z_3=3$ , alors:

  $\;\;y_1 \preccurlyeq z_1$ , $\; z_2 \preccurlyeq y_2$, $\;\; y_3 \preccurlyeq z_3$ donc $(y_1,y_2,y_3)\preccurlyeq (z_1,z_2,z_3)$

 Or nous avons: $\;\;x_1 \preccurlyeq z_1$ , $\; z_2 \preccurlyeq x_2$, $\;\; z_3 \preccurlyeq x_3$ donc $(z_1,z_2,z_3)\preccurlyeq (x_1,x_2,x_3)$           et non pas $(x_1,x_2,x_3)\preccurlyeq (z_1,z_2,z_3)$
   
 $\Rightarrow$ Donc cette relation de préférence n'est pas transitive.
