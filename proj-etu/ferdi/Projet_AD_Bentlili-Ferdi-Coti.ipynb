On se place sur $X ={R}^3_+$ et on considère la relation de préférence notée 
$\preccurlyeq$  définie comme  suit:

$(x_1,x_2,x_3)\preccurlyeq(y_1,y_2,y_3)$ si et seulement pour au moins deux valeurs de $i$ dans {1, 2, 3} on a $x_i \leq y_i$.
    
# **$\Bbb1.\Bbb1)$**
#**Montrons que cette relation de préférence est complète.**


---



L'on est sur ${R}^3_+$ , telle que la relation de préférence, ici, envisagée est :
$(x_1,x_2,x_3)\preccurlyeq(y_1,y_2,y_3)$, si et seulement si pour 2 valeurs de $i$ $\epsilon{\{1,2,3\}}$ on ai : $x_i \leq y_i$.


***Définition*** : Une relation de préférence est dite complète si tous les éléments de notre ensemble X sont comparables.
Soient $\forall(x,y)$ $\epsilon{X^2}$ : x $\preccurlyeq$ y , ou , y $\preccurlyeq$ x.
\
Cette relation de préference est ainsi dite transistive si $\forall(x,y,z)$ $\epsilon{X^3}$, x $\preccurlyeq$ y et y $\preccurlyeq$ z 
\
alors x $\preccurlyeq$ z.

Suivant la relation $\leq$ , l'on a $\forall(x,y)$ $\epsilon{X}$ et $\forall$ $i$ $\epsilon{\{1,2,3\}}$ nous avons soit $x_i\leq y_i$ soit $x_i\geq y_i$. \
Ainsi cette relation est donc définie dans les cas suivant : \
\
$\ast$ Cas 1 : si $x_1 \leq y_1$  et $x_2 \leq y_2$ $\Rightarrow$ $x_3 \leq y_3$ ou $y_3 \leq x_3$ \
\
$\ast$Cas 2 : si $x_2 \leq y_2$  et $x_3 \leq y_3$ $\Rightarrow$ $x_1 \leq y_1$ ou $y_1 \leq x_1$ \
\
$\ast$Cas 3 : si $x_3 \leq y_3$  et $x_1 \leq y_1$ $\Rightarrow$ $x_2 \leq y_2$ ou $y_2 \leq x_2$ 

Et si $x_i=y_i$ alors $\forall$ $i$ ; x equivaut à y (soit $x_i$ est préférer à $y_i$ et $y_i$ à $x_i$) ainsi il n'y aucunes préférences. \\


\
# **$\Bbb1.\Bbb2)$**
#**À l'aide d'un contrexemple, montrons que cette relation de préférence n'est pas transitive.** 


---


\
Dans notre situation, ( $\forall(x_i,y_i)$ $\epsilon$ X ) si l'on prend par exemple : $x_1\leq y_1$ et $x_2\leq y_2$, cela n'implique pas que $x_3\leq y_3$.
\
Au vue de la définition de la relation $\leq$, cet exemple est une éventualité mais on peut aussi avoir : $x_1\leq y_1$; $x_2 \leq y_2$ et $y_3 \leq x_3$.
\
Ce qui montre que cette relation de préférence n'est pas transitive.
\
De plus montrons cela de façon algébrique : soient : x =(1,1,2) , y =(1,0,2) , z =(0,0,2).
\
Nous obtenons bien que $x\preccurlyeq y$ et $y \preccurlyeq z$, $\forall$ $z \preccurlyeq x$.

```python
import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """

class PbResolution(Exception):
    """Probleme de resolution du LP
    """

# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 (toutes les mains sont supposées équiprobables)
 
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
        g = g + (1/l) * gmfun(s1, s2, Mains[i])
    return g

A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
        A[i, j] = gfun(S1[i], S2[j])

d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte

# Calcul de la valeur du jeu en stratégies mixtes (minimax)
c =-np.ones(n) # min (c|x) = -max (-c|x)
b = np.ones(m) # Ax <= b 
bounds = (0, None) # 0 <= x

method='simplex'
#method='revised simplex'
#method='interior-point'

res = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method=method)
if not res.success: raise PbResolution

v = 1 / (-res.fun)
x = v * res.x
v = v - d
xx = np.round_(x, decimals=8)

print('v = ', v)
print('x = ', xx)
print('')
print("Stratégie mixte optimale de la joueuse 2 :")
for j in range(0, n):
    if xx[j] > 0:
        print(S2[j], "avec proba", xx[j])
        
```
