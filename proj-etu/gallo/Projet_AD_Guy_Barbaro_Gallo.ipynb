On se place sur $X = {\mathbf{R}}^3_+$ et on considère la relation de préférence notée $\preccurlyeq$ définie comme suit : 
$(x_1,x_2,x_3) \preccurlyeq (y_1,y_2,y_3)$ si et seulement pour au moins deux valeurs de $i$ dans $\{1, 2, 3\}$ on a $x_i \leq y_i$.

### 1.1
Montrer cette relation de préférence est complète.

Une relation de préférence $\preccurlyeq$ est dite complète si tous les éléments de X sont comparables : \
    $(\forall (x,y)$ $\in X^2 = X * X)$ : \
    $x \preccurlyeq y$ ou $y \preccurlyeq x$

Analysons les différents cas sur $X = R^3+$ : 
- $x1 \leq y1$   $x2 \leq y2$    $x3 \leq y3$ d'où $X \preccurlyeq Y$ 
- $x1 \leq y1$   $x2 \leq y2$    $y3 \leq x3$ d'où $X \preccurlyeq Y$ 
- $x1 \leq y1$   $y2 \leq x2$    $y3 \leq x3$ d'où $Y \preccurlyeq X$ 
- $y1 \leq x1$   $y2 \leq x2$    $y3 \leq x3$ d'où $Y \preccurlyeq X$ 

On peut donc toujours comparer les éléments de X entre eux. Donc la relation de préférence $\preccurlyeq$ est complète.


### 1.2
À l'aide d'un contrexemple, montrer que cette relation de préférence n'est pas transitive.

Soit $\preccurlyeq$ une relation de préférence, cette relation est transitive si : \
    $(\forall (x,y,z)$ $\in X^3$ : $x \preccurlyeq y$  et $y \preccurlyeq z$ alors $x \preccurlyeq z$ \

Ici , X = (x1,x2,x3) Y = (y1,y2,y3) Z=(z1,z2,z3)
Si $X \preccurlyeq Y$ et $ Y\preccurlyeq Z$ , montrons que l'on a pas toujours $X \preccurlyeq Z$. \

Prenons donc un contrexemple : \

X = (x1 =4, x2=3, x3=33) Y = (y1=5, y2=1, y3=37) Z = (z1=6, z2 = 2, z3 = 32)\
On a : \
$x1 \leq y1$ et $x3 \leq y3$ et donc $X \preccurlyeq Y$. \
$y1 \leq z1$ et $y2 \leq2$ donc $Y \preccurlyeq Z$.\
$z2 \leq x2$ et $z3 \leq x3$ donc $Z \preccurlyeq X$. \

On a donc bien montré, que $X \preccurlyeq Y$ et $Y \preccurlyeq Z$ n'implique pas toujours $X \preccurlyeq Z$.
Donc, cette relation n'est pas transitive.
