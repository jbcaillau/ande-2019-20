###Exercice 1:

1.1) Sachant que $(x_1,x_2,x_3)$ $\preccurlyeq$ $(y_1,y_2,y_3)$ si et seulement pour deux valeurs de i $\in$ {1,2,3} on a $x_i$ $\leq$ $y_i$\
L'axiome de complétude vérifie que la relation de préférence est complète.
C'est-à-dire que pour tout (x,y)$\in$ $X^2$
$$ X \preccurlyeq Y \quad et \quad Y \preccurlyeq X $$
On pose X = $(x_1,x_2,x_3)$ et Y = $(y_1,y_2,y_3)$

Cette relation est donc complète dans les cas où:

- $x_1 \le y_1 \quad et \quad x_2 \le y_2 \quad et \quad (x_3 \le y_3 \quad ou \quad x_3 \ge y_3)$
- $x_1 \le y_1 \quad et \quad (x_2 \le y_2 \quad ou \quad x_2 \ge y_2) \quad et \quad x_3 \le y_3$
- $(x_1 \le y_1 \quad ou \quad x_1 \ge y_1) \quad et \quad x_2 \le y_2 \quad et \quad x_3 \le y_3$

$ \quad \mathit{ou}$
- $x_1 \ge y_1 \quad et \quad x_2 \ge y_2 \quad et \quad (x_3 \le y_3 \quad ou \quad x_3 \ge y_3)$
- $x_1 \ge y_1 \quad et \quad (x_2 \le y_2 \quad ou \quad x_2 \ge y_2) \quad et \quad x_3 \ge y_3$
- $(x_1 \le y_1 \quad ou \quad x_1 \ge y_1) \quad et \quad x_2 \ge y_2 \quad et \quad x_3 \ge y_3$

La relation de préférence est donc bien complète car le consommateur est toujours capable de choisir entre X et Y ou de les estimer comme étant équivalent.

1.2) Une relation est dite transitive si pour 3 paniers (X,Y,Z),\
Si $$X \preccurlyeq Y \quad et \quad Y \preccurlyeq Z \quad alors \quad X \preccurlyeq Z$$

On pose donc un nouveau panier $Z=(z_1,z_2,z_3)$

Prenons donc X=(2,2,1) Y=(0,1,0) et Z=(0,1,1)\
On a $X \preccurlyeq Y$ et $ Y \preccurlyeq Z$ par definition de complétude 
Mais ici $X \not\preccurlyeq Z$

Cette relation de préférence n'est pas transitive par ce contre-exemple.
