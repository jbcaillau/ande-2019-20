## Exercice 1
On se place sur $X = {\mathbf{R}}^3_+$ et on considère la relation de préférence notée $\preccurlyeq$ définie comme suit : 
$(x_1,x_2,x_3) \preccurlyeq (y_1,y_2,y_3)$ si et seulement pour au moins deux valeurs de $i$ dans $\{1, 2, 3\}$ on a $x_i \leq y_i$.

### 1.1
Montrer cette relation de préférence est complète.

**Réponse :**
Soit $A$ le groupe de biens $(x_1,x_2,x_3)$ et $B$ le groupe de biens $(y_1,y_2,y_3)$. Cette relation de préférence ne peut etre complète que si et seulement si le consommateur soit toujours capable de dire s'il préfère $A$ à $B$ ou $B$ à $A$ ou si $A$ et $B$ sont équivalents.

On peut distinguer deux cas: $x_i \leq y_i$ pour au moins deux valeurs de $i$ dans ${1, 2, 3}$ et $x_i \leq y_i$ pour au maximum une valeur de $i$ dans ${1, 2, 3}$. Dans le premier cas, on aura donc que $A \preccurlyeq B$ et dans le second $B \preccurlyeq A$, donc cette relation de préférence est complète.


### 1.2
À l'aide d'un contrexemple, montrer que cette relation de préférence n'est pas transitive.

**Réponse :**
Soit $C$ le groupe de biens  $(z_1,z_2,z_3)$ et soit  $\preccurlyeq$ la même relation de préférence complète vu à la question précédente. $B \preccurlyeq C$ avec $y_i \leq z_i$ pour au moins deux valeurs de $i$ dans ${1, 2, 3}$ et $A \preccurlyeq B$  avec $x_i \leq y_i$ pour au moins deux valeurs de $i$ dans ${1, 2, 3}$ Soit $C \preccurlyeq A$ avec $z_i \leq x_i$ pour au moins deux valeurs de $i$ dans ${1, 2, 3}$. Si cette relation de préférence était transitive, on aurait eu $A \preccurlyeq B \preccurlyeq C$ et $A \preccurlyeq C$ et non pas  $A \preccurlyeq B \preccurlyeq C$ et $C \preccurlyeq A$ ce qui montre que cette relation de préférence n'est pas transitive.

Discuter du maximum de la fonction d'utilité $U(x,y)=3x^3 + x^2 y - 17x^2 - 4xy + 17x - y + 5$ sous les contraintes $x\geq 0$, $y\geq 0$ et la contrainte de revenu $5x+3y\leq r$ pour $r=20$ puis pour $r=5$.

$U(x,y)=3x^3+x^2y−17x^2−4xy+17x−y+5$  avec  $x≥0$ ,  $y≥0$  avec la contrainte de revenu  $5x+3y≤r$  pour  $r=20$ et $r=5$.

Ici, c'est un problème de maximisation. On cherche le Maximum de  $U(x,y)$ 

Soit  $u_0≥0$  un niveau d'utilité fixé, il faut définir tout d'abord la courbe d'indifférence : On cherche Y, en fonction de X.

$U(x,y)=u_0<=>y(x^2−4x−1)=−3x^3+17x^2−17x−5+u_0$   
$<=>y=f(x)=\frac{−3x^3+17x^2−17x−5+u_0}{x^2−4x−1}$ 

$\frac{∂f(x)}{∂x}=\frac{−3x^4+24x^3−42x^2−xu_0−24x+4u−3}{(x2−4x−1)^2} <0$ 

$\frac{∂f(x)}{∂x^2}=\frac{2u(3x^2−12x+17)}{(x^2−4x−1)^3} >0$ 
 

Grace à la dérivée seconde, on en déduit que les préférences sont convexes. On en déduit le $TMS=−f′(x)>0$

On étudie  $r=20$ , notre contrainte est :  $5x+3y=20<=>y=\frac{20−5x}{3}$ 

On sait que  $y=\frac{−3x^3+17x^2−17x−5+u_0}{x^2−4x−1}=\frac{20−5x}{3}$ 

Cherchons les points critiques pour trouver le maximum de la fonction d'utilité :  $y=\frac{20−5x}{3}=0<=>x=4$ 

Pour  $x=4,y=0$ , on respecte bien la contrainte car  $5∗4+3∗0=20$ , le couple  $(4,0)$  maximise donc bien la fonction d'utilité.

Si  $r=5$ , notre contrainte est :  $5x+3y=5<=>y=\frac{5−5x}{3}$ 

On sait que  $y=\frac{−3x^3+17x^2−17x−5+u_0}{x^2−4x−1}=\frac{5−5x}{3}$ 

Cherchons les points critiques pour trouver le maximum de la fonction d'utilité.  $y=\frac{5−5x}{3}=0<=>x=1$ 

Pour  $x=1,y=0$ , on respecte bien la contrainte car  $5∗1+3∗0=5$ , le couple  $(1,0)$  maximise donc bien la fonction d'utilité.

### 3.6 (BONUS)
Reprendre les questions 3.1 à 3.5 en supposant le nouveau croupier honnête, de sorte que toutes les combinaisons de mains sont équiprobables.

````python
import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """

class PbResolution(Exception):
    """Probleme de resolution du LP
    """

# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 (toutes les mains sont supposées équiprobables)
 
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
        g = g + (1/l) * gmfun(s1, s2, Mains[i])
    return g

A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
        A[i, j] = gfun(S1[i], S2[j])

d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte

# Calcul de la valeur du jeu en stratégies mixtes (minimax)
c =-np.ones(n) # min (c|x) = -max (-c|x)
b = np.ones(m) # Ax <= b 
bounds = (0, None) # 0 <= x

method='simplex'
#method='revised simplex'
#method='interior-point'

res = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method=method)
if not res.success: raise PbResolution

v = 1 / (-res.fun)
x = v * res.x
v = v - d
xx = np.round_(x, decimals=8)

print('v = ', v)
print('x = ', xx)
print('')
print("Stratégie mixte optimale de la joueuse 2 :")
for j in range(0, n):
    if xx[j] > 0:
        print(S2[j], "avec proba", xx[j])
````
