### 3.2
Déterminer les stratégies _prudentes_ de chaque joueuse, c'est à dire la stratégie $\max_i \min_j$ pour la joueuse 1 et $\min_j \max_i$ de la joueuse 2. Y a-t-il équilibre (en stratégies pures) ?

L'idée de john Von Neumam pour retrouver une situation d'équilibre consiste à jouer un grand nombre de parties en adaptant un point de vue probabiliste. $J_1$ va jouer la stratégie $i=1$ avec la probabilité $y_1$, la stratégie $i=2$ avec une probabilité $y_2$ etc...
De même $J_2$ jouera la stratégie j appartient à {${1,..,n}$} avec la proba $x_j$. On parle alors de stratégie "mixte" et on voit que cela revient à arrondir $S_1$ (respectivement $S_2$) en $Somme_m$ := {y appartient à $R^m$ | $y \geq 0$, $y_1+...+y_m = 1$} (respectivement pour x)
Choisir une stratégie mixte $y=(y_1,...,y_m)$ appartient à $Somme_m$ revient donc à fixer avec quelle probabilité $y_1$ $J_1$ jouera $i=1$

Le résultat de Nash (1950) donne que si un jeu avec un nombre de joueurs fini,  et que tous les joueurs ont un nombre de stratégie finie, alors ce jeu possède au moins un équilibre. Si un jeu n'a pas d'équilibre en stratégies pures alors il doit avoir un équilibre en stratégies mixtes. 

Ici, dans cet exercice, il y a un nombre fini de joueur (2) et chaque joueur possède un nombre de stratégie finie ($3^3=27$ pour $J_1$ et $4^3=64$ pour $J_2$) donc ce jeu possède au moins un équilibre.

````python
#code pour l'ex 3A questions 3.1 --> 3.5 #
import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """

class PbResolution(Exception):
    """Probleme de resolution du LP
    """

# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 (la main (2,3) a une probabilité égale à (1/4) de sortir, les autres ont une probabilité égale à (3/20).
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
      if i == 3:
        g = g +(1/4) * gmfun(s1, s2, Mains[i])
      else : 
        g = g + (3/20) * gmfun(s1, s2, Mains[i])
    return g

A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
        A[i, j] = gfun(S1[i], S2[j])

d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte

# Calcul de la valeur du jeu en stratégies mixtes (minimax)
c =-np.ones(n) # min (c|x) = -max (-c|x)
b = np.ones(m) # Ax <= b 
bounds = (0, None) # 0 <= x

method='simplex'
#method='revised simplex'
#method='interior-point'

res = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method=method)
if not res.success: raise PbResolution

v = 1 / (-res.fun)
x = v * res.x
v = v - d
xx = np.round_(x, decimals=8)

print('v = ', v)
print('x = ', xx)
print('')
print("Stratégie mixte optimale de la joueuse 2 :")
for j in range(0, n):
    if xx[j] > 0:
        print(S2[j], "avec proba", xx[j])
````

````python
# Code pour la question 3.6 #
import numpy as np
from scipy.optimize import linprog

# Exceptions
class StrategieInconnue(Exception):
    """Strategie Inconnue
    
    Attribut :
        s -- stratégie
    """
    def __init__(self, s):
        self.s = s

class MainInconnue(Exception):
    """Main Inconnue
    
    Attribut :
        m -- main
    """
    def __init__(self, m):
        self.m = m

class MainVide(Exception):
    """Main Vide
    """

class PbResolution(Exception):
    """Probleme de resolution du LP
    """

# Stratégies de la joueuse 1
S1m = [ 'M', 'PV', 'PP' ] # M = Mise, P = Passe, V = Voit
l1 = len(S1m)
S1 = [ (S1m[i], S1m[j], S1m[k]) for i in range(0, l1) for j in range(0, l1) for k in range(0, l1) ]
m = len(S1)

# Stratégies de la joueuse 2
S2m = [ 'VP', 'VM', 'PP', 'PM' ] # M = Mise, P = Passe, V = Voit
l2 = len(S2m)
S2 = [ (S2m[i], S2m[j], S2m[k]) for i in range(0, l2) for j in range(0, l2) for k in range(0, l2) ]
n = len(S2)

# Mains pouvant être distribuées
Mains = [ (1,2), (1,3), (1,4),(2,1), (2,3), (2,4), (3,1), (3,2), (3,4), (4,1), (4,2), (4,4) ] # main 1 < main 2 < main 3

# Calcul des gains
def gmfun(s1, s2, main):
    """g = gmfun(s1, s2, main)
    Gain de la joueuse 1 en fonction de la main distribuée aux deux joueuses.

    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    main : main (= cartes) distribuées aux joueuses
    """
    if s1 not in S1: raise StrategieInconnue(s1)
    if s2 not in S2: raise StrategieInconnue(s2)
    if main not in Mains: raise MainInconnue(main)

    s1m = s1[main[0]-1]
    s2m = s2[main[1]-1]

    if s1m[0] is 'M': # la joueuse 1 mise

        if s2m[0] is 'V': # la joueuse 2 voit
            if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
            else: gm = -2
        elif s2m[0] is 'P': # la joueuse 2 passe
            gm = 1 # la joueuse 1 emporte le pot
        else: raise StrategieInconnue(s2)

    elif s1m[0] is 'P': # la joueuse 1 passe

        if s2m[1] is 'P': # la joueuse 2 passe
            if main[0] > main[1]: gm = 1 # la meilleure main emporte le pot
            else: gm = -1
        elif s2m[1] is 'M': # la joueuse 2 mise
            if s1m[1] is 'V': # la joueuse 1 voit
                if main[0] > main[1]: gm = 2 # la meilleure main emporte le pot
                else: gm = -2
            elif s1m[1] is 'P': # la joueuse 1 passe
                gm = -1
            else: raise StrategieInconnue(s1)
        else: raise StrategieInconnue(s2)

    else: raise StrategieInconnue(s1)

    return gm

def gfun(s1, s2):
    """g = gfun(s1, s2)
    Gain en espérance de la joueuse 1 (toutes les mains sont supposées équiprobables)
 
    s1 : triplet de stratégies de la joueuse 1
    s2 : triplet de stratégies de la joueuse 2
    """
    l = len(Mains)
    if l is 0: raise MainVide
    g = 0.
    for i in range(0, l):
        g = g +(1/l) * gmfun(s1, s2, Mains[i])
    return g

A = np.zeros((m, n))
for i in range(0, m):
    for j in range(0, n):
      A[i, j] = gfun(S1[i], S2[j])


d = abs(np.min(A)) + 1
A = A + d # translate les gains pour garantir une valeur > 0 du jeu mixte

# Calcul de la valeur du jeu en stratégies mixtes (minimax)
c =-np.ones(n) # min (c|x) = -max (-c|x)
b = np.ones(m) # Ax <= b 
bounds = (0, None) # 0 <= x

method='simplex'
#method='revised simplex'
#method='interior-point'

res = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method=method)
if not res.success: raise PbResolution

v = 1 / (-res.fun)
x = v * res.x
v = v - d
xx = np.round_(x, decimals=8)

print('v = ', v)
print('x = ', xx)
print('')
print("Stratégie mixte optimale de la joueuse 2 :")
for j in range(0, n):
    if xx[j] > 0:
        print(S2[j], "avec proba", xx[j])
````
